```python
#handling outliers with IQR

Q1= df['Salary'].quantile(0.25)
Q3= df['Salary'].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

df[(df['Salary']<lower) | (df['Salary']>upper)]
```

```python
#I am keeping this outlier as it is important

df['Salary']= np.where(df['Salary']> upper, upper, df['Salary'])
```

```python
#Feature Selection- Drop Age column

df=df.drop(columns=['Age'])
```

```python
#Handling Categorical data for further analysis and modeling

df=pd.get_dummies(df, drop_first=True)
```

```python
x= df.drop('Salary', axis=1)
y=df['Salary']
```

```python
#Train-Test-Split

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test= train_test_split(
    x,y, test_size=0.2, random_state=42)
```

```python
#Build the model

from sklearn.linear_model import LinearRegression

model= LinearRegression()
model.fit(x_train, y_train)
```

```python
#Predict

y_pred= model.predict(x_test)
```

```python
#Evaluation

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
#Improve model using Random Forest

from sklearn.ensemble import RandomForestRegressor

rf= RandomForestRegressor(random_state=42)
rf.fit(x_train, y_train)
```

```python
y_pred_rf= rf.predict(x_test)
```

```python
print("MAE:", mean_absolute_error(y_test, y_pred_rf))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_rf)))
print("R2 Score:", r2_score(y_test, y_pred_rf))
```

```python
#Feature Importance

coeff=pd.Series(model.coef_, index=x.columns)
coeff.sort_values(ascending=False)
```

```python
coeff.sort_values().plot(kind='barh')
plt.title("Feature Importance")
plt.show()
```
